### GraphRoots Figure Reproducibility

In our project we use NetworkX and Pyvis to show connections between people and places. We have created a way to find information on someone in Grassroots database. For that person, we create an individualized and interactive relationship graph. Surrounding them are nodes representing other organizations and people with respective labels. An edge indicates that they are connected to each other. If you hover over the edge, information is displayed on their connection including where and when they worked together. By running this file you can produce each of the graphs used in our final project.

To create the figures in our README.md and final report, you will need to follow our INSTALL.md to run the frontend of GraphRoots. Once you have it up and running and open the website, you will find Contact Lookup on the left bar. Click it and enter 'lis' and take a screenshot to make the figure.
![Image](Images/Individual_Search_Bar.png)

Click the search suggestion for 'Lisa Blackburn' when it pops up. Once you click it, take a screenshot to recreate our figure.
This page has the contact information and connections for the searched person.
![Image](Images/Individual_Search.png)

Click the GENERATE CONTACT GRAPH on this page to recreate our next graph which is the Individual Graph. Take a screenshot to recreate the figure. This figure has Lisa Blackburn in the center with lines connecting to companies and organizations she is currently or was affiliated with in the past. Connected to those are more nodes representing people who she has worked or been affiliated with in those organizations. The size of the nodes represent the number of years that the two have had a connection.
![Image](Images/Individual_Graph.png)

For the next figure first click in the greyed out region to exit the graph. Click Business/Company Lookup and input 'arc' into the search bar and take a screenshot.
![Image](Images/Business_Search_Bar.png)

Click the suggestion "Archer-Patel (Construction) when it pops up. Take a screenshot to recreate the figure.
![Image](Images/Business_Search.png)

Click on GENERATE BUSINESS GRAPH. Take a screenshot to recreate the figure. At the center is Archer-Patel and connected are current and past employees designated by blue and red respectively. Connected to those people are their current and past employment and affiliations.
